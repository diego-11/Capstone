```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
# Lee el archivo y crea el DataFrame
df = pd.read_csv('online_gaming_behavior_dataset.csv')

# Muestra las primeras 5 filas
print(df.head())
```

```python
# 2. Escalar los datos para que tengan la misma importancia
scaler = StandardScaler()
datos_escalados = scaler.fit_transform(df)

# 3. Definir el número de clústeres (K) e inicializar K-means
kmeans = KMeans(n_clusters=3, random_state=42)

# 4. Ajustar el modelo a los datos
kmeans.fit(datos_escalados)

# 5. Obtener las etiquetas de los clústeres asignados a cada fila
df['Cluster'] = kmeans.labels_
```
